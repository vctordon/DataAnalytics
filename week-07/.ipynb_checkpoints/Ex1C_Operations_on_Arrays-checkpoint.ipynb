```python
print(r_4)
```

```python

np.random.shuffle(r_4)
print(r_4)
```

```python
# it does not mix individual values across the entire array, only the order of rows changes.
```

```python

new_shuffle = np.random.permutation(r_4)

print(new_shuffle)
print(r_4)
```

```python
# Permutation is used when you want a shuffled copy without changing the original
```

## Grocery Problem

```python
# Order: vegetables, fruits, snacks, beverages
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
print(f"the first bag has {np.sum(bag1)} items, and the second one has {np.sum(bag2)} items.")
```

```python
totals = bag1+bag2
print(f"There are {totals[0]} vegtables, {totals[1]} fruits,{totals[2]} boxes of snacks, and {totals[3]} beverages.")
```

## Teamwork

```python
ali = np.array([5,4,6,4,2])
ben = np.array([0,5,5,3,4])   # missed Monday
carla = np.array([3,6,0,3,4]) # missed Wednesday
```

```python

all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python

print(np.sum(all_knocks, axis=1))  # per person
print(np.sum(all_knocks))          # total group
```

```python

daily_totals = np.sum(all_knocks, axis=0)
rolling_total = np.cumsum(daily_totals)

estimated_votes = rolling_total * 0.5
print(estimated_votes)
```
