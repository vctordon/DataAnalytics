## Advanced Plotting Techniques

```python

import numpy as np
import matplotlib.pyplot as plt

```

```python

products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python

plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Sales by Product')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python

sizes = np.random.rand(50) * 100

plt.scatter(x, y, s=sizes, c=sizes, cmap='viridis')
plt.title('Scatter Plot with Variable Marker Sizes')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.colorbar()
plt.show()
```

```python
## Subplots
```

```python

# Create data
x = np.linspace(0, 10, 100)
y = np.sin(x)

random_data = np.random.randn(1000)

# Create subplots
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# First subplot: Line chart
ax[0].plot(x, y)
ax[0].set_title('Sine Wave')
ax[0].set_xlabel('X values')
ax[0].set_ylabel('sin(X)')

#Second subplot: Histogram
ax[1].hist(random_data, bins=30)
ax[1].set_title('Histogram of Random Data')
ax[1].set_xlabel('Value')
ax[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
