```python
import numpy as np
import pandas as pd
```


## Series


```python
aa = [1, 2, 3, 4]                      # list
ab = (5, 6, 7, 8)                      # tuple
ac = {9, 10, 11, 12}                   # set
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}  # dictionary
```

```python

ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac))
ser4 = pd.Series(ad)

print(ser1)
print(ser2)
print(ser3)
print(ser4)

```

## DataFrames from Series


```python

df1 = pd.DataFrame(ser1)
df1
```

```python

df1.insert(1, "col2", ser2)
df1.insert(2, "col3", ser3)

df1
```

```python

df1["col4"] = ser4
df1
```


# ser4 comes from a dictionary, so it uses labels (keys) as index.
# The DataFrame index is numeric (0,1,2...), so they don’t match.
# This causes NaN values to appear because the indexes don’t align.
